Graph: Top 5 Selling Product Categories (using Seaborn)


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Category': ['Electronics', 'Clothing', 'Home & Kitchen', 'Books', 'Beauty', 'Toys', 'Groceries'],
    'Total_Sales': [150000, 120000, 95000, 85000, 60000, 40000, 30000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Sort top 5 categories by sales
top_categories = df.sort_values(by='Total_Sales', ascending=False).head(5)

# Plotting
plt.figure(figsize=(10, 6))
sns.barplot(x='Category', y='Total_Sales', data=top_categories, palette='viridis', hue='Category')  # Apply color palette with hue
plt.title('Top 5 Selling Product Categories in E-Commerce')
plt.xlabel('Product Category')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=45)
plt.tight_layout()

# Save plot image (optional)
plt.savefig('top_categories_bar_chart.png')

# Show plot
plt.show()
```

## Top 5 Selling Product Categories

This bar chart visualizes the top 5 product categories by total sales in a mock e-commerce dataset.

Libraries used:
- Seaborn
- Matplotlib
- Pandas

Key Insight:
Helps understand which product lines are the most profitable for a business.
